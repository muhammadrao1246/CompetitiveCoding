```python
import re

def isPalindrome(s : str):
    filtered = s.lower()
    filtered = re.sub("[^a-z0-9]", "", filtered)
    
    for i in range(0, len(filtered)):
        last = len(filtered) - i - 1
        
        if filtered[i] != filtered[last]:
            return False
        
        if (len(filtered) - i - 1) - i <= 1:
            return True
        
        
isPalindrome("A man, a plan, a canal: Panama")
```
