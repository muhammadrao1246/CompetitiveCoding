```python
#O(n)
def canJump(nums: list[int]) -> bool:
        n = len(nums)

        jump = n-1
        i = n-2
        while i >= 0:
            print(i, jump)
            if i + nums[i] >= jump:
                jump = i 
            i-=1
        
        return jump == 0
    
print(canJump([0,1]))
```
