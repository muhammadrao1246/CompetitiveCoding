```python
def func(haystack, needle):
    """
    :type haystack: str
    :type needle: str
    :rtype: int
    """
    if haystack == needle: 
        return 0
    for i in range(0, len(haystack) - len(needle)+1):
        currSubStr =  haystack[i: i + len(needle)]
        if currSubStr == needle:
            return i
    return -1


# func("  to   the moon  fl", "fly")
func(" f", "f")
func("a", "a")
```
