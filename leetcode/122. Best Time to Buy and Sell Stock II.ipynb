```python
import sys
def maxProfit(prices: list[int]) -> int:
    minPrice = prices[0]
    totalProfit = 0
    nextProfit = 0
    n = len(prices)
    for i in range(1, n):
        minPrice = min(prices[i], minPrice)
        if prices[i] > minPrice:
            nextProfit = prices[i] - minPrice
            print(minPrice, nextProfit)
            minPrice = prices[i]
            totalProfit += nextProfit

    return totalProfit

print(maxProfit([7,1,5,3,6,4]))    
```
