```python
def maxProfit(prices: list[int]) -> int:
    firstMinPrice = prices[0]
    secondMinPrice = prices[0]
    firstMaxProfit = 0
    secondMaxProfit = 0
    
    n = len(prices)
    # atmost two transactions may be done and cannot buy another stock before selling the previous one
    for i in range(1, n):
        secondMinPrice = min(prices[i], secondMinPrice)
        secondMaxProfit = max(firstMaxProfit, prices[i] - firstMinPrice)
        
        firstMinPrice = min(prices[i], firstMinPrice)
        firstMaxProfit = max(firstMaxProfit, prices[i] - firstMinPrice)
    
print(maxProfit([3,3,5,0,0,3,1,4]))   
```
