```python
def lengthOfLastWord(s):
        """
        :type s: str
        :rtype: int
        """
        lastWord = ""
        newWord = ""
        for i in range(0, len(s)):
            if s[i] != " ":
                newWord += s[i]
                lastWord = newWord
            elif s[i] == " ":
                newWord = ""
        
        return len(lastWord)
lengthOfLastWord("  fly me   to   the moon  ")
```
