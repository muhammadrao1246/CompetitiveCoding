```python
# O(n) - optimized
def romanToInt( s): 
        """
        :type s: str
        :rtype: int
        """
        # mapping roman to integer
        singleMap = {"I":1, "V":5, "X":10, "L":50, "C":100, "D":500, "M":1000}
        total = 0
        i = 0
        while i < len(s):
            curr = s[i]
            next = (s[i+1] if i+1<len(s) else "")
            if curr == "I" and next == "V":
                total += 4
            elif curr == "I" and next == "X":
                total += 9 
            elif curr == "X" and next == "L":
                total += 40 
            elif curr == "X" and next == "C":
                total += 90 
            elif curr == "C" and next == "D":
                total += 400 
            elif curr == "C" and next == "M":
                total += 900
            else:
                total += singleMap[curr]
                i+=1
                continue
            i+=2
        
        return total
romanToInt("MCMXCIV")
```

```python
# O(n) -  Easiest way to think in terms of map
def romanToInt( s): 
        """
        :type s: str
        :rtype: int
        """
        # mapping roman to integer
        singleMap = {"I":1, "V":5, "X":10, "L":50, "C":100, "D":500, "M":1000}
        joinedMap = {"IV":4, "IX":9, "XL":40, "XC":90, "CD":400, "CM":900}
        total = 0
        i = 0
        while i < len(s):
            two_numeral = s[i]+(s[i+1] if i+1<len(s) else "" )
            if two_numeral in joinedMap:
                print(f"{two_numeral} -> {joinedMap[two_numeral]}")
                total += joinedMap[two_numeral]
                i+=2
            else:
                one_numeral = s[i]
                total += singleMap[one_numeral]
                print(f"{one_numeral} -> {singleMap[one_numeral]}")
                i+=1
        
        return total
romanToInt("MCMXCIV")
```
