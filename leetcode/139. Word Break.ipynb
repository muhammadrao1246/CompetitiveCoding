```python
def wordBreak( s: str, wordDict: list[str]) -> bool:
        
        for item in wordDict:
            if item in s:
                s = s.replace(item, "")  
        
        if len(s) == 0:
            return True 
        return False 

print(wordBreak("catsanddog", ["cats","dog","sand","and","cat"]))
```
