```python
# Binary search tree from Array
# Definition for a binary tree node.
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right
    
    def __str__(self):
        def performINORDERDFS(head, newArr: list):
            if head is None: return
            
            if head.left is not None: performINORDERDFS(head.left, newArr)
            newArr.append(head.val)
            if head.right is not None: performINORDERDFS(head.right, newArr)
        def performPREORDERDFS(head, newArr: list):
            if head is None: return
            
            newArr.append(head.val)
            if head.left is not None: performPREORDERDFS(head.left, newArr)
            if head.right is not None: performPREORDERDFS(head.right, newArr)
        def performPOSTORDERDFS(head, newArr: list):
            if head is None: return
            
            if head.left is not None: performPOSTORDERDFS(head.left, newArr)
            if head.right is not None: performPOSTORDERDFS(head.right, newArr)
            newArr.append(head.val)
            
        # going DFS to get sorted array
        head = self 
        st = ""
        newArr = []
        performINORDERDFS(head, newArr)
        st += "INORDER: "+str(newArr)+"\n"
        performPREORDERDFS(head, newArr)
        st += "PRE: "+str(newArr)+"\n"
        performPOSTORDERDFS(head, newArr)
        st += "POST: "+str(newArr)+"\n"
        return st

# Divide and Conquer breaking down problems into sub problems
def ArrayToBST(nums, start, offset):
    if start > offset: return None
    midpoint = int(start + (offset - start)/2) # if start is 5 and offset 10 so 10-5/2 = 2.5 we have to plus start again to make it midpoint
    head = TreeNode(nums[midpoint], None, None)
    head.left = ArrayToBST(nums, start, midpoint-1)
    head.right = ArrayToBST(nums, midpoint+1, offset)
    # print(f"HEAD: {head.val} LEFT: {head.left.val} RIGHT: {head.right.val}")
    return head
    
def sortedArrayToBST(nums):
        """
        :type nums: List[int]
        :rtype: Optional[TreeNode]
        """
        return ArrayToBST(nums, 0, len(nums)-1)
        # BELOW CODE IS NOT WORKING FOR LEETCODE cause does not produce balanced so building with recursion
        # we are going to pick midpoint of array index so the tree stay balanced on both sides
        midValueIndex = int(len(nums)/2)
        head = TreeNode(nums[midValueIndex], None, None)

        # inserting each element one by one except the midValueIndex
        for i in range(0, len(nums)):
            if i == midValueIndex: continue
            val = nums[i]
            currHead = head
            while currHead:
                # if its smaller than currHead go in left
                if val < currHead.val:
                    if currHead.left is None:
                        currHead.left = TreeNode(val, None, None)
                        break
                    currHead = currHead.left
                elif val >= currHead.val:
                    if currHead.right is None:
                        currHead.right = TreeNode(val, None, None)
                        break
                    currHead = currHead.right
        
        return head

                    
print(sortedArrayToBST([0,1,2,3,4,5]))
```
