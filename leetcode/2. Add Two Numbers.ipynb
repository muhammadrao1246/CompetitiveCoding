```python
# Definition for singly-linked list.
class ListNode:
    def __init__(self, x):
        self.val = x
        self.next = None
    
    def __str__(self):
        return f"{self.val}"
    
    def displayFullList(self):
        st = str(self.val)
        head = self.next 
        while head is not None:
            st += " -> " + str(head.val)
            head = head.next
        return st
```

```python
# Cycled linked list
linklistONE = ListNode(9)
linklistONE.next = ListNode(9)
linklistONE.next.next = ListNode(9)


linklistTWO = ListNode(9)

def addTwoNumbers(l1:ListNode, l2:ListNode) -> ListNode:
        """
        :type l1: Optional[ListNode]
        :type l2: Optional[ListNode]
        :rtype: Optional[ListNode]
        """
        new_list = None
        new_list_pointer = new_list
        next_carry = 0
        one = l1
        two = l2
        while one or two:
            # total sum of both digits
            sum_val = (one.val if one else 0) + (two.val if two else 0)
            # sum plus carry if any
            current_val = sum_val+next_carry
            # next carry if current_val >= 10 else 0
            next_carry = int(current_val/10) if current_val >= 10 else 0
            # adding node to new list
            if new_list is None: 
                new_list = ListNode(current_val%10)
                new_list_pointer = new_list
            else: 
                new_list_pointer.next = ListNode(current_val%10)
                new_list_pointer = new_list_pointer.next
            print(f"SUM: {sum_val} SAVED: {new_list_pointer.val} CURRENT: {current_val} NEXT_CARRY: {next_carry}")
            one = one.next if one else None
            two = two.next if two else None
            print(l1.displayFullList())
            print(l2.displayFullList())
            
        if next_carry > 0:
            new_list_pointer.next = ListNode(next_carry)

        return new_list

print(addTwoNumbers(linklistONE, linklistTWO).displayFullList())
```
