```python
def InfixToPostfix(s:str):
    isRightToLeftAssoc = lambda x : x == "^" 
    
    getPrec = lambda operator : 3 if operator == "^" else (2 if operator in "/*" else (1 if operator in "+-" else -1))
        
    ops = []
    new = ""
    for i in s:
        # if its an operator
        if i == "(":
            ops.append(i)
        elif i == ")":
            # we have to remove all operators from stack and attach them to the expression string
            while ops and ops[-1] != "(":
                new += ops.pop()
            ops.pop()
        elif i.isalnum():
            new += i # attach operands
        else: # means operators
            # so we have to take the operators in or out based on their precedence
            while ops and ops[-1] != "(" and (getPrec(i) < getPrec(ops[-1]) or (getPrec(i) == getPrec(ops[-1]) and not isRightToLeftAssoc(i))):
                new += ops.pop()
            ops.append(i)
        
        print(f"{i} -> {ops} ----- {new}")
    # just popping out any remaining operator
    while ops:
        new += ops.pop()
        
    return new

# InfixToPostfix("a+b+c")
InfixToPostfix("(((a^2/((b-c)+d))*(e-a))*c)")
```

```python
def InfixToPrefix(s:str):
    # first we have to reverse the string
    new = ""
    for i in reversed(s):
        if i == "(":
            new += ")"
        elif i == ")":
            new += "("
        else:
            new += i
                
    print(new)
    # second run postfix
    s = InfixToPostfix(new)
    # reverse again
    new=""
    for i in reversed(s):
        new += i
    return new

# InfixToPostfix("a+b+c")
InfixToPrefix("(((a^2/((b-c)+d))*(e-a))*c)")
```
