# Selection Sort

```python
# IN SELECTION SORT
# AN ELEMENT IS SELECTED AND THEN COMPARED WITH OTHER ALL ELEMENTS TO CHECK FOR THE MINIMUM OF THEM ALL TO BE REPLACED WITH
# O(n2) - Inplace - Space Complexity O(1)
def Selection_Sort(arr: list):
    print(arr)
    for i in range(0, len(arr)):
        for j in range(i + 1, len(arr)):
            if arr[i] > arr[j]:
                arr[i], arr[j] = arr[j], arr[i]
    
    return arr

print(Selection_Sort([64, 25, 12, 22, 11]))  
```

```python
# Bubble Sort

# All index elements are compared and swapped with each number of len(arr) times
# Inplace O(1) - Complexity O(n2)

def Bubble_Sort(arr: list):
    for i in range(0, len(arr)):
        for j in range(0, len(arr)):
            if arr[i] < arr[j]:
                arr[i], arr[j] = arr[j], arr[i]
                
    return arr 

print(Bubble_Sort([11,0,64, 34, 25, 12, 22, 11, 90]))
```

```python
# Insertion Sort
# We will set each element to its best position by comparing than all previous elements
# we do need to consider that index 0 is a sorted list
# O(n2) - Inplace O(1)
def Insertion_Sort(arr: list):
    for i in range(1, len(arr)): # starting from index 1 as previous list will be sorted cause have [ELEM] 1 element in it
        # current value
        key = arr[i] # On Index 1 -> 90
        # Sorted list till back of the indexes
        j = i - 1
        
        # iterate in opposite direction in sorted list to place KEY in its appropriate position
        # swappings are possible with O(1) best and O(n) in worst case
        while j >= 0 and arr[j] > key:
            arr[j + 1] = arr[j] # swapping with right lesser item to make the list sorted
            arr[j] = key # key will be compared again and again with the current arr[j] and will be swapped
            j -= 1 # go back
            
    return arr

Insertion_Sort([100, 90, 80, 70, 60, 50])        
```
