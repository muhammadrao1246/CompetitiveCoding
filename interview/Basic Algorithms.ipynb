# Fibonnaci Calculator  


```python
# Fibonnaci using recursion
# 0 1 1 2 3 5 8 (sequence)
# 0 1 2 3 4 5 6 (index order)
# O( exponential greater than time not suitable)
def fibonacci(n):
    if n == 0: return 0
    if n == 1: return 1    
    return fibonacci(n-1)+fibonacci(n-2)

print(fibonacci(3))

# Fibonnaci using DP
def fibonacci_with_loop(n):
    if n == 0: return 0
    if n == 1: return 1
    temp = 0
    a = 0
    b = 1
    for i in range(2, n+1):
        temp = a + b
        a = b
        b = temp        
    
    return temp   


# print(fibonacci_with_loop(3))     
```

# Brackets Validate 

```python
# validate (((())))

def validate_brackets(st):
    count = 0
    
    for s in st:
        if s == "(":
            count += 1
        else:
            count -= 1
    
    if count == 0:
        return "Valid"
    
    return "Invalid"    
            
    
print(validate_brackets("((()))"))

# validating brackets with stack
def validate_brackets_with_stack(string):
    stack = [] # push when open found pop when close found
    for s in string:
        if s == "(" or s == "{" or s == "[":
            stack.append(s)
        else: # if close found but stack empty 
            if stack == []:
                return "Invalid"
            elif stack[len(stack)-1] in ["(", "[", "{"] and s in [")", "]", "}"]:
                stack.pop()
        print(stack)
    if stack == []:
        return "Valid"
    return "Invalid"
print(validate_brackets_with_stack('(((){}(([])[])))'))
```
