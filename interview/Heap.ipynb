```python
# Heap Tree must follow ACBT
# Building Heaps Two Methods
# 1. Build Heap or Create Heap O(nlog(n)) - Insert each element one by one
# 2. Heapify O(n) - Build all elements at once
# Heap Sort O(nlog(n)) - deleting and rearranging using any appraoch

# 1. Create or Build Heap builds the heap tree one by one
new_tree = []
def build_heap(tree: list, newElement):
    tree.append(newElement)
    n = len(tree)
    i = n - 1
    
    # comparing with its parents
    while i > 0:
        parent = int((i-1) / 2)
        if tree[parent] < tree[i]: # swapping if that happens
            tree[parent], tree[i] = tree[i], tree[parent]
        else: # means no need modifications
            break
        
        i = parent # go to parent to parent to parent 
    
    return tree
    

# Tree Printing Lib
from binarytree import build
# Random Tree
tree = [23,7,43,600,432,123,500]

# executing build_heap O(nlog(n)) first to go to each then traverse through its height
# for elem in tree:
#     new_tree = build_heap(new_tree, elem)
#     print(build(new_tree))
    

# 2. Heapify O(n) - Inplace - Post Order Follow
def heapify(arr, i, n):
    if n <= 1: return arr
    
    left = i * 2 + 1
    right = i * 2 + 2

    if right >= n and left >= n: # means at leaf node
        # print(f"LEAF DETECTED: {arr[i]}")
        return i
    
    largest = i
    if right < n: 
        temp = heapify(arr, right, n)
        largest = largest if arr[largest] > arr[temp] else temp
    if left < n: 
        temp = heapify(arr, left, n)
        largest = largest if arr[largest] > arr[temp] else temp 

    if arr[largest] > arr[i]:
        arr[i], arr[largest] = arr[largest], arr[i]
        heapify(arr, largest, n) # go down deep again to reswap new swpapped nodes with their childrens
        # print(build(arr))
    
    return i    

# print(build(tree))
heapify(tree, 0, len(tree))
print(build(tree))

# 3. Deleting Elements from Heap - O(nlog(n)) - Inplace 
def deleting_from_top_one_by_one(arr: list, n: int):
    # total_deleted_nodes
    n -= 1
    if n <= 0: return
    # replacing most right leaf node with the root node
    arr[0], arr[n] = arr[n], arr[0]
    # # calling heapify O(n)
    print(build(arr))
    heapify(arr, 0, n)
    print(build(arr))
    
    deleting_from_top_one_by_one(arr, n)

# PRODUCE HEAP SORT O(nlog(n)) array inplace
# print(build(tree))
deleting_from_top_one_by_one(tree, len(tree))
print(tree)
print(build(tree))
```
