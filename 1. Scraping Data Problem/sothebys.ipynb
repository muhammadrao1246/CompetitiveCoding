```python
import cloudscraper

# https://www.sothebys.com/api/auth0login?lang=en&fromHeader=Y&locale=en
# Create a scraper instance
scraper = cloudscraper.create_scraper(browser='firefox')

# Make a request to a Cloudflare-protected site
response = scraper.get('https://www.sothebys.com/api/auth0login?lang=en&fromHeader=Y&locale=en')
print(f"Status code: {response.status_code}")
```
