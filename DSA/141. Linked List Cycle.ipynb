```python
# Definition for singly-linked list.
class ListNode:
    def __init__(self, x):
        self.val = x
        self.next = None
    
    def __str__(self):
        return f"{self.val}"
    
    def displayFullList(self):
        st = str(self.val)
        head = self.next 
        while head is not None:
            st += " -> " + str(head.val)
            head = head.next
        return st
        
# Cycled linked list
linklist = ListNode(1)
linklist.next = ListNode(2)
linklist.next.next = ListNode(3)
linklist.next.next.next = ListNode(4)
linklist.next.next.next.next = ListNode(5)
linklist.next.next.next.next.next = linklist.next
        
# SIMPLE SET BASED SOLUTION O(n)    
def hasCycle(head: ListNode):
        recordNode = set()  
        while head:
            print(head.val)
            if head in recordNode:
                return True
            recordNode.add(head)
            head = head.next
        
        return False
hasCycle(linklist)
```

```python
# A Relatively fast solution using turtoise and rabbit approach 
def hasCycle( head: ListNode):
        
        if head:
            if head.next is None:
                return False
        turtle = head
        hare = head
        while turtle and hare:
            hare = hare.next.next if hare.next else head
            turtle = turtle.next
            print(f"Hare {hare} Turtle {turtle}")
            if hare == turtle:
                return True

        return False
    
hasCycle(linklist)
```
