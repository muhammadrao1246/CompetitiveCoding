```python
# Definition for singly-linked list.
class ListNode:
    def __init__(self, x):
        self.val = x
        self.next = None
    
    def __str__(self):
        return f"{self.val}"
    
    def displayFullList(self):
        st = str(self.val)
        head = self.next 
        while head is not None:
            st += " -> " + str(head.val)
            head = head.next
        return st
        
# Cycled linked list
linklist = ListNode(1)
linklist.next = ListNode(2)
linklist.next.next = ListNode(3)
linklist.next.next.next = ListNode(4)
linklist.next.next.next.next = ListNode(5)

linklistTwo = ListNode(6)
linklistTwo.next = ListNode(7)
linklistTwo.next.next = linklist.next.next


        
# O(N) operation
def getIntersectionNode(headA: ListNode, headB: ListNode):
        """
        :type head1, head1: ListNode
        :rtype: ListNode
        """
        listNodeSet = set()
        print(headA.displayFullList())
        print(headB.displayFullList())
        while headA or headB:
            if headB: 
                if headB in listNodeSet:
                    return headB
                listNodeSet.add(headB) 
                headB = headB.next
            if headA: 
                if headA in listNodeSet:
                    return headA
                listNodeSet.add(headA)
                headA = headA.next


print(getIntersectionNode(linklist, linklistTwo))
            
```
