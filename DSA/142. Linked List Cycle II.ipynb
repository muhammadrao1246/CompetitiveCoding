```python
# Definition for singly-linked list.
class ListNode:
    def __init__(self, x):
        self.val = x
        self.next = None
    
    def __str__(self):
        return f"{self.val}"
    
    def displayFullList(self):
        st = str(self.val)
        head = self.next 
        while head is not None:
            st += " -> " + str(head.val)
            head = head.next
        return st
        
# Cycled linked list
linklist = ListNode(1)
linklist.next = ListNode(2)
linklist.next.next = ListNode(3)
linklist.next.next.next = ListNode(4)
linklist.next.next.next.next = ListNode(5)
linklist.next.next.next.next.next = linklist.next
```

```python
# Using Set or Dict method O(n)
def detectCycle(head):
        """
        :type head: ListNode
        :rtype: ListNode
        """
        nodeSet = set()
        currHead = head
        while currHead:
            if currHead in nodeSet:
                return currHead
            nodeSet.add(currHead)
            currHead = currHead.next

        return None

print(detectCycle(linklist))
```

```python
# Using Two pointer approach
def detectCycle(head : ListNode):
        hare = head
        turtle = head
        
        # if only one node then also throw None
        if head is not None and head.next is None:
            return None
        
        while turtle and hare:
            # hare jumping two
            hare = hare.next.next if hare.next is not None else None
            # jumping one turtle
            turtle = turtle.next
            # if turtle or hare is None that means no cycle
            if turtle is None or hare is None:
                return None
            if hare == turtle: break
            
        hare = head # reseting fast pointer this time they are going to meet at start of the cycle somehow
        while hare != turtle:
            hare = hare.next
            turtle = turtle.next
            
        return hare


# # Cycled linked list [1, 2] 
linklist = ListNode(1)
linklist.next = ListNode(2)
linklist.next.next = linklist


# # Cycled linked list
linklist = ListNode(3)
linklist.next = ListNode(2)
linklist.next.next = ListNode(0)
linklist.next.next.next = ListNode(-4)
linklist.next.next.next.next = linklist.next

print(detectCycle(linklist))
```
